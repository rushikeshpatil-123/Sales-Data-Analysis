```python
# Import Data Manipulation Libraries
import pandas as pd
import numpy as np

# Import Data Visiualization Libraries
import matplotlib.pyplot as plt
import seaborn as sns

#filter Out Warning Messages
import warnings
warnings.filterwarnings(action = "ignore")
```

```python
# Load Dataset

df = pd.read_csv("/content/sales_data_sample.csv", encoding='ISO-8859-1')

# Display first 5 records
print("Dataset Preview")

df.head()


# Dataset Information

print("\n dataset Information ")
print(df.info())

#Basic Statistics
print("\n Sales Statistics")
print("Total Sales:", df['SALES'].sum())
print("Average Sales:", df['SALES'].mean())
print("Maximum Sales:",df['SALES'].max())
print("Minimum Sales:", df['SALES'].min())
```

```python
#Region/Country Wise Sales
country_sales = df.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=False)

print("\n Top 20 Countries by Sales")
print(country_sales.head(10))
```

```python
# Top Products
product_sales = df.groupby('PRODUCTLINE')['SALES'].sum().sort_values(ascending=False)

print("\n Top Selling Product Categories")
print(product_sales)
```

```python
# Bar Chart
plt.figure(figsize = (10,6))
product_sales.plot(kind = "bar")
plt.title("Sales by product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
# Pie Chart

plt.figure(figsize=(8,8))
product_sales.plot(kind='pie', autopct='%1.1f%%')
plt.ylabel("")
plt.title("Sales Distribution by Product Line")
plt.show()
```

```python
# Top 5 Products
top_products = df.groupby('POSTALCODE')['SALES'].sum().sort_values(ascending=False).head(5)
print("\nTop 5 Performing Products")
print(top_products)

# Top 5 Countries
top_countries = country_sales.head(5)

plt.figure(figsize=(8,5))
top_countries.plot(kind='bar')
plt.title("Top 5 Countries by Sales")
plt.xlabel("Country")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```
